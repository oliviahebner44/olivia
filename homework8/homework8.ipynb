```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def make_sine_wave(x, A, w):
    y = A * np.sin(w * x)
    return y

x = np.linspace(0, 2*np.pi, 1000)

amplitude = [0.5, 1, 1.5, 2, 2.5]
frequency = [1, 2, 3, 4, 5]

fig, axes = plt.subplots(1, 3, figsize=(25, 5))

for A, w in zip(amplitude, frequency):
    # first subplot
    axes[0].plot(x, make_sine_wave(x, A, 1), label=f'A = {A}')

    # second subb
    axes[1].plot(x, make_sine_wave(x, 1, w), label=f'w = {w}')

    # Third subpp
    axes[2].plot(x, make_sine_wave(x, A, w),
                 label=f'A = {A}, w = {w}')

# titles 
axes[0].set_title('Varying Amplitude', fontsize = 16)
axes[1].set_title('Varying Frequency', fontsize = 16)
axes[2].set_title('Varying Amplitude and Frequency', fontsize = 16)

# x and y labels:
for ax in axes:
    ax.set_xlabel('x', fontsize=13)
    ax.set_ylabel('y', fontsize=13)
    ax.legend(fontsize=11)

plt.tight_layout()
plt.show()
```
