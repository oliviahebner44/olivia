```python
# 4 Prime Star System 
arr = np.array([
    [2, 3, 5],
    [14, 6, 8],
    [11, 13, 17],
    [7, 10, 13]
])

def is_prime(n):
    if n < 2:
        return False 

    for i in range(2, n):
        if n % i ==0:
            return False 

    return True

def prime_rows(arr):
    rows = []

    for row in arr:
        for num in row:
            if is_prime(num):
                rows.append(row)
                break 

    return np.array(rows) 


print(prime_rows(arr))
```

```python
checkerboard = np.array(board)
```

```python
def rev_checkerboard():
    board = make_board()
    board[0::2, 0::2] = 0
    board[0::2, 1::2] = 1
    board[1::2, 0::2] = 1
    board[1::2, 1::2] = 0
    return board 
```

```python
print(rev_checkerboard())
```

```python
# 6 The Expanding Universe 
def expansion(universe, n):
    result = []

    for word in universe:
        result.append((" " * n).join(word))

    return np.array(result)
```

```python
universe = np.array(['galaxy', 'clusters'])
print(expansion(universe, 1))
print(expansion(universe, 2)) 
```

```python
# 7 Second Brightest Star 
def secondim(arr):
    sorted_arr = np.sort(arr, axis = 1)
    return sorted_arr[:, -2]
```

```python
np.random.seed(123)
stars = np.random.randint(500, 2000, (5,5))

print(secondim(stars))
```
