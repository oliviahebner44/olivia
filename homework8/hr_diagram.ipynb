```python
import numpy as np
import matplotlib.pyplot as plt 

data = np.loadtxt("../../course_assignments/homework7/star_data.txt", skiprows=1)
```

```python
data.shape
```

```python
# 2642 rows, 2 columns
magnitude = data[:, 0]
temperature = data[:, 1]

print(np.argmin(magnitude), np.min(magnitude))
print(np.argmax(magnitude), np.max(magnitude))
```

```python
print(np.argmin(temperature), np.min(temperature))
print(np.argmax(temperature), np.max(temperature))
```

```python
brightest = np.argmin(magnitude)
dimmest = np.argmax(magnitude)
hottest = np.argmax(temperature)
coldest = np.argmin(temperature)
print(brightest)
print(dimmest)
```

```python
log_magnitude = np.log(magnitude)
log_temperature = np.log(temperature) 

plt.figure(figsize=(8,6))

plt.scatter(log_temperature, log_magnitude, c=temperature, cmap='coolwarm', label ='stars')

plt.colorbar(label='Temperature (K)')

plt.scatter(log_temperature[brightest], log_magnitude[brightest], color='black', marker='*', s=300, label='Brightest Star')

plt.scatter(log_temperature[dimmest], log_magnitude[dimmest], color='purple', marker='*', s=300, label='Dimmest Star')

plt.scatter(log_temperature[hottest], log_magnitude[hottest], color='green', marker='*', s=300, label='Hottest Star')

plt.scatter(log_temperature[coldest], log_magnitude[coldest], color='orange', marker='*', s=300, label='Coldest Star')

plt.xlabel('Log Temperature')
plt.ylabel('Log Magnitude')
plt.title('Hertzsprung-Russell Diagram')

plt.gca().invert_xaxis()
plt.gca().invert_yaxis()

plt.show()
```
