```python
# homework 7 
# Pandas DataFrame

import pandas as pd
```

```python
df = pd.read_csv("../../course_assignments/homework8/state_data.csv")
print(df.shape)
# 52 rows, 4 columns
```

```python
df.head() #columns are state, latitude, longitude, and name. first state is Alaska, last state is Wyoming, 51 total entries, 
```

```python
df.tail()
```

```python
median_lat = df["latitude"].median()
median_lon = df["longitude"].median()

print(median_lat)
print(median_lon)
```

```python
df["lat_dist"] = abs(df["latitude"] - median_lat)
df["lon_dist"] = abs(df["longitude"] - median_lon)
```

```python
print(df.shape)
```

```python
close_lat = df["lat_dist"].idxmin()
print(f"Closest to median latitude: {df.loc[close_lat, 'name']} by {df.loc[close_lat, 'lat_dist']:.2f} degrees")
```

```python
far_lat = df["lat_dist"].idxmax()
print(f"Farthest to median latitude: {df.loc[far_lat, 'name']} by {df.loc[far_lat, 'lat_dist']:.2f} degrees")
```

```python
close_lon = df["lon_dist"].idxmin()
print(f"Closest to median longitude: {df.loc[close_lon, 'name']} by {df.loc[close_lon, 'lon_dist']:.2f} degrees")
```

```python
far_lon = df["lon_dist"].idxmax()
print(f"Farthest to median longitude: {df.loc[far_lon, 'name']} by {df.loc[far_lon, 'lon_dist']:.2f} degrees")
```

```python
# Seaborn
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = sns.load_dataset('flights')
```

```python
df.head()
```

```python
df = df.dropna(axis=1)
```

```python
df.shape # 144 rows, 3 columns)
```

```python
print("first five rows:") 
display(df.head())
print("last five rows:")
display(df.tail())
```

```python
df.columns
```

```python
fig, axes = plt.subplots(2, 2, figsize=(12, 8))

# left Scatterplot of year vs passengers
sns.scatterplot(data=df, x='year', y='passengers', color='blue', ax=axes[0, 0])
axes[0, 0].set_title('Year vs Passengers')
axes[0, 0].set_xlabel('Year')
axes[0,0].set_ylabel('Passangers (thousands)')

# right Scatterplot of month vs passengers
sns.scatterplot(data=df, x='month', y='passengers', color='red', ax=axes[0, 1])
axes[0, 1].set_title('Month vs Passengers')
axes[0, 0].set_xlabel('Month')
axes[0,0].set_ylabel('Passangers (thousands)')

# left histogram
sns.histplot(data=df, x='passengers', color='green', ax=axes[1, 0])
axes[1, 0].set_title('Distribution of Passengers')
axes[0,0].set_ylabel('Passangers (thousands)')
axes[0, 0].set_xlabel('Frequency')

# right boxplot
sns.boxplot(data=df, x='month', y='passengers', color='orange', ax=axes[1, 1])
axes[1, 1].set_title('Passengers by Month')
axes[0, 0].set_xlabel('Month')
axes[0,0].set_ylabel('Passangers (thousands)')

plt.tight_layout()
plt.show()
```

```python
# top left is a scatterplot of year vs passangers to show the relationship over time.
# top right is scatterplot of number vs passangers to see the variations over the year.
# bottom left is a histogram of pagssangers to show the frequency of different passanger counts across the data 
# bottom right is a boxplot comparing the distribution of passanger counts across the year. 
```
