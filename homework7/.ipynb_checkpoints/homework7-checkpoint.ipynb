```python
# homework 7 
# Pandas DataFrame

import pandas as pd
```

```python
df = pd.read_csv("../../course_assignments/homework8/state_data.csv")
print(df.shape)
# 52 rows, 4 columns
```

```python
df.head() #columns are state, latitude, longitude, and name. first state is Alaska, last state is Wyoming, 51 total entries, 
```

```python
df.tail()
```

```python
median_lat = df["latitude"].median()
median_lon = df["longitude"].median()

print(median_lat)
print(median_lon)
```

```python
df["lat_dist"] = abs(df["latitude"] - median_lat)
df["lon_dist"] = abs(df["longitude"] - median_lon)
```

```python
print(df.shape)
```

```python
close_lat = df["lat_dist"].idxmin()
print(f"Closest to median latitude: {df.loc[close_lat, 'name']} by {df.loc[close_lat, 'lat_dist']:.2f} degrees")
```

```python
far_lat = df["lat_dist"].idxmax()
print(f"Farthest to median latitude: {df.loc[far_lat, 'name']} by {df.loc[far_lat, 'lat_dist']:.2f} degrees")
```

```python
close_lon = df["lon_dist"].idxmin()
print(f"Closest to median longitude: {df.loc[close_lon, 'name']} by {df.loc[close_lon, 'lon_dist']:.2f} degrees")
```

```python
far_lon = df["lon_dist"].idxmax()
print(f"Farthest to median longitude: {df.loc[far_lon, 'name']} by {df.loc[far_lon, 'lon_dist']:.2f} degrees")
```
